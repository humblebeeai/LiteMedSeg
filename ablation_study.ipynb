# Ablation Study on CVC-ClinicDB dataset

```python
import torch
from model_ablation_study import *

if torch.cuda.is_available():
    device = torch.device("cuda")
    print("CUDA device found.")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
    print("MPS device found.")
else:
    device = torch.device("cpu")
    print("Using CPU.")
```

## CVC-ClinicDB dataset

```python
# Dataset Paths
data_path = "data/"
dataset_name = "CVC-ClinicDB"

if  dataset_name == "CVC-ClinicDB":
    image_dir = "Original"
    mask_dir = "Ground Truth"
elif dataset_name == "Kvasir-SEG":
    image_dir = "images"
    mask_dir = "masks"

dataset_path = os.path.join(data_path, dataset_name)

# Image Preprocessing Parameters
IMG_SIZE = (256, 256)

dataset = MedicalImageSegmentationData(
    dataset_path=dataset_path,
    image_dir=image_dir,
    mask_dir=mask_dir,
    img_size=IMG_SIZE,
    batch_size=8,
    test_size=0.1,
    random_state=42,
    binarize_mask=True,
)

train_loader, test_loader = dataset.dataloaders()

print(f"Total number of samples: {len(dataset)}")
print(f"Number of training samples: {len(train_loader.dataset)}")
print(f"Number of testing samples: {len(test_loader.dataset)}")

# dataset.visualize_samples()
```

## wandb use

```python
use_wandb = True

if use_wandb:
    from dotenv import load_dotenv
    load_dotenv()
    import wandb
    wandb.login(key=os.getenv("WANDB_API_KEY"))
```

## run ablation study

```python
training_results, test_results = run_complete_ablation_study(
    train_loader, 
    test_loader, 
    test_loader, 
    base_channels=8, 
    dataset_name=dataset_name,
    epochs=1, 
    device='cuda', 
    base_save_path="./ablation_study",
    use_wandb=use_wandb
)
```
